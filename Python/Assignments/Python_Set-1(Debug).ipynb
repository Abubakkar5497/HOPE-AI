7.	Scenario: A retail store offers a 20% discount if a customer’s total order exceeds $100. Write logic to calculate the final amount to be paid after applying the discount.

```python
amount = float(input('Enter total order value: $'))

if amount>=100:
    discount = amount*0.20
    Total_mount = amount - discount
    print(f"Discount of 20% applied: {discount}")
else:
    discount = 0
    Total_amount = amount
    print('No Discount Applied')

print(f"Total Amount after discount: ${Total_mount}")
```

8.	Scenario: A banking system processes withdrawal requests and ensures the user has enough balance.

 Write logic to check if a user has enough balance before allowing a withdrawal and update the remaining balance accordingly.


```python
Withdraw_amount = int(input('Enter your withdrawal amount: '))
balance = int(input('Enter your Balance amount: '))

if (balance >= Withdraw_amount):
    remaining_balance = Withdraw_amount-balance
    print('Your Bank acoount having enough balance so your withdrawal is processed')

else:
    remaining_balance = balance
    print('Your Bank acoount doesnt have enough balance so your withdrawal is cancelled')

print(f"Your Remaining Balance is: ${remaining_balance}")
    
```

9. A calendar system verifies whether a given year is a leap year based on standard leap year rules.
 Write logic to determine whether a given year is a leap year.


```python
import calendar

year = int(input('Enter the year: '))
if calendar.isleap(year):
    print(f"{year} is leap year")
else:
    print(f"{year} is not leap year")
```

10.	Scenario: A program filters out only even numbers from a given list.
 Write logic to extract and return only the even numbers from a list.


```python
number = [12,13,7,15,18,16,20,21]
even_number = []
for num in number:
    if ((num%2==0)):
        even_number.append(num)

print('Even Number is : ', even_number)
        
            
```
