```python
list = [1,2,3,4,5,6,7]


def subfields():
    for Subfields in list:
        if(Subfields==1):
            print('Sub-fields in AI are:')
        elif(Subfields==2):
            print('Machine Learning')
        elif(Subfields==3):
            print('Neural Network')
        elif(Subfields==4):
            print('Vision')
        elif(Subfields==5):
            print('Robotics')
        elif(Subfields==6):
            print('Speech processing')
        else:
            print('Natural Language Processing')
            
```

```python
subfields()
```

```python
def OddEven():
    num = int(input('Enter a Number: '))
    if((num%2)==0):
        print(f'{num} is Even number')
    else:
        print(f'{num} is Odd number')
```

```python
OddEven()
```

```python
def EligibilityForMarriage(gender, age):
    if gender == "Male":
        if age >= 21:
            print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")
    elif gender == "Female":
        if age >= 18:
            print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")


```

```python
print("Your Gender: Male")
print("Your Age: 20")
EligibilityForMarriage('Male', 20)
```

```python
def FindPercentage():
    subject1 = 90
    subject2 = 87
    subject3 = 95
    subject4 = 95
    subject5 = 93
    
    total = subject1 + subject2 + subject3 + subject4 + subject5
    percentage = (total / 500) * 100
    
    print(f"Subject1= {subject1}")
    print(f"Subject2= {subject2}")
    print(f"Subject3= {subject3}")
    print(f"Subject4= {subject4}")
    print(f"Subject5= {subject5}")
    print(f"Total: {total}")
    print(f"Percentage: {percentage}")
```
