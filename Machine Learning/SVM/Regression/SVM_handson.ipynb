```python
import pandas as pd
import numpy as np
```

```python
dataset = pd.read_csv("50_Startups.csv")
```

```python
dataset.head()
```

```python
dataset = pd.get_dummies(dataset,drop_first=True)
dataset = dataset.astype(int)
```

```python
dataset.head()
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend',
       'State_Florida', 'State_New York']]
dependent = dataset[['Profit']]
```

```python
independent.head()
```

```python
dependent.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(independent, dependent, test_size = 0.30, random_state=0)
```

```python
#Standerdization i am going because to improve my model perforamnce
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
from sklearn.svm import SVR
regressor = SVR(kernel='linear', C=0.01) # rbf is radial bais function, Poly,sigmoid,linear
regressor.fit(X_train,y_train) # Model is created with this code
```

```python
y_pred = regressor.predict(X_test)
```

```python
bais = regressor.intercept_
```

```python
bais
```

```python
regressor.n_support_
```

```python
regressor.support_
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
```
