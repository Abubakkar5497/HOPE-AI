```python
import pandas as pd
```

```python
dataset = pd.read_csv('insurance_pre.csv')
```

```python
dataset.head()
```

```python
dataset
```

```python
dataset.columns
```

```python
dataset = pd.get_dummies(dataset)
dataset = dataset.astype(int)
```

```python
dataset.head()
```

```python
dataset.columns
```

```python
Independent = dataset[['age', 'bmi', 'children','sex_female', 'sex_male',
       'smoker_no', 'smoker_yes']]
Dependent = dataset[['charges']]
```

```python
Independent.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(Independent,Dependent,test_size = 0.30,random_state = 42)
```

```python
# Linear regression

from sklearn.linear_model import LinearRegression
regressor1 = LinearRegression()
regressor1 = regressor1.fit(X_train,y_train)
```

```python
y_pred = regressor1.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
```

```python
r_score
```

```python
# SVR Regression

from sklearn.svm import SVR
regressor2 = SVR(kernel='sigmoid', C =5000)
regressor2 = regressor2.fit(X_train,y_train)
y_pred = regressor2.predict(X_test)
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```

```python
# Decision Tree Regression

from sklearn.tree import DecisionTreeRegressor

regressor3 = DecisionTreeRegressor(criterion='poisson',splitter='best', max_features=None, random_state=0)
regressor3 = regressor3.fit(X_train,y_train)
y_pred = regressor3.predict(X_test)
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```

```python
from sklearn.ensemble import RandomForestRegressor

regressor4 = RandomForestRegressor(n_estimators=100,criterion='poisson',random_state=42,max_features=None)
regressor4 = regressor4.fit(X_train,y_train)
y_pred = regressor4.predict(X_test)
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```
