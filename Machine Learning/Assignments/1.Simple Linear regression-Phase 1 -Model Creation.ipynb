```python
import pandas as pd # imports the Pandas library in Python and gives it the alias pd, allowing to use its data analysis functions like pd.DataFrame() or pd.read_csv()
import numpy as np # imports the NumPy library in Python and gives it the alias np, allowing you to use its numerical and array operations like np.array() or np.mean()
```

```python
dataset = pd.read_csv("Salary_Data.csv") # from pd class we calling function read_csv() and assign to dataset variable
```

```python
dataset # assigned variable
```

```python
dataset.tail() # show last 5 rows of the dataset
```

```python
#Independent and Dependent variable seprate
Independent_variable = dataset[["YearsExperience"]] # we taking years experience as independent so assign to Independent_variable
Dependent_variable = dataset[["Salary"]] #we taking Salary as dependent so assign to dependent_variable
```

```python
Independent_variable.tail()
```

```python
#Split Train and Test
from sklearn.model_selection import train_test_split #sklearn is the library and sklearn.model_selection is the model and import train_test_split is the function use for splitng training and test data.
X_train,X_test,y_train,y_test = train_test_split(Independent_variable,Dependent_variable,test_size=0.30,random_state=0)
# X_train is 70% of inpute data for training
#X_test is 30% of input data for testing
#y_train is 70% of output data for training
#y_test is 30% of output data for testing
#test_size=0.30 is test size is 30% of the data
#random_state=0 is ensure  reproducibility of the code again we run the code we will get same output
```

```python
#Model creation
from sklearn.linear_model import LinearRegression # sklearn.linear_model is model and LinearRegression is function
regressor = LinearRegression() # assign LinearRegression function to regressor variable
regressor.fit(X_train,y_train) # .fit is method or function in python and we training the model X_train- 70% input data and y_train 30% test data
```

```python
weight=regressor.coef_ # weight is calculated in coef_ and assign this weight or slope to weight variable
weight
```

```python
bais = regressor.intercept_ # bais is stored in intercept_ and assign bais to bais variable
bais
```

```python
# Evaluation Metrix
y_pred = regressor.predict(X_test) # predict is method bcz the function in the object we use the trainned model regressor and predict 30% input data and assign to y_pred variable
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score # r2_score is function and used for check the performance of the mode.
r_score = r2_score(y_test,y_pred) # y_test is actual data and y_pred is predicted data and comparing this two withr2_score and check the accuracy of the mode.
```
