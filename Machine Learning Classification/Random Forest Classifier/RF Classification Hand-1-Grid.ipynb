```python
#import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
```

```python
dataset.head()
```

```python
dataset = pd.get_dummies(dataset,drop_first = True)
dataset=dataset.astype(int)
```

```python
dataset.head()
```

```python
dataset = dataset.drop('User ID',axis=1)
```

```python
dataset.head()
```

```python
dataset['Purchased'].value_counts()
```

```python
dataset.columns
```

```python
ind = dataset[['Age', 'EstimatedSalary', 'Gender_Male']]
dep = dataset[['Purchased']]
```

```python
ind.shape
```

```python
#split train,test,split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(ind,dep,test_size=1/3,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV



param_grid = {'criterion':['gini','entropy'],
              'max_features': ['auto','sqrt','log2'],
              'n_estimators':[10,100]} 



grid = GridSearchCV(RandomForestClassifier(), param_grid, refit = True, verbose = 3,n_jobs=-1,scoring='f1') 
   
# fitting the model for grid search 
grid_clas=grid.fit(X_train, y_train) 
 
```

```python
#print(grid.best_params_) 
re=grid.cv_results_
#print(re)
grid_predictions = grid.predict(X_test) 
```

```python
print(re)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,grid_predictions)
print(cm)
```

```python
from sklearn.metrics import classification_report
clr = classification_report(y_test,grid_predictions)
print(clr )
```

```python
import pickle
filename = 'Finalize_Model_RF.sav'
pickle.dump(grid_clas,open(filename, 'wb'))
pickle.dump(sc, open("Scaler_RF.sav", 'wb'))
```
