```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset.head()
```

```python
X = dataset.iloc[:,[3,4]].values
```

## 1. AffinityPropagation

```python
from sklearn.cluster import AffinityPropagation
affpro = AffinityPropagation()
label = affpro.fit_predict(X)
```

```python
supervised1 = pd.DataFrame(dataset)
```

```python
supervised1['Clusters']=label
```

```python
supervised1.head()
```

```python
aff = sns.lmplot(data = supervised1, x = supervised1.columns[3], y = supervised1.columns[4], hue=supervised1.columns[5],fit_reg=False,legend=True,legend_out=True)
```

## 2.Mean Shift

```python
from sklearn.cluster import MeanShift
ms = MeanShift(bandwidth = 25)
label2 = ms.fit_predict(X)
```

```python
supervised2 = pd.DataFrame(dataset)
```

```python
supervised2['Clusters']=label2
```

```python
supervised2.head()
```

```python
ms = sns.lmplot(data = supervised2, x = supervised2.columns[3], y = supervised2.columns[4], hue = supervised2.columns[5],fit_reg=False, legend=True,legend_out = True)
```

## Spectral clustering

```python
from sklearn.cluster import SpectralClustering
sc = SpectralClustering(n_clusters=10)
label3 = sc.fit_predict(X)
```

```python
supervised3 = pd.DataFrame(dataset)
supervised3['Clusters']=label3
supervised3.head()
```

```python
sc = sns.lmplot(data = supervised3,x=supervised3.columns[3],y=supervised3.columns[4],hue=supervised3.columns[5],fit_reg = False,legend=True,legend_out=True)
```

## DBSCAN

```python
from sklearn.cluster import DBSCAN
db = DBSCAN(eps=0.5)
label4 = db.fit_predict(X)
```

```python
supervised4 = pd.DataFrame(dataset)
supervised4['Clusters']=label4
supervised4.head()
```

```python
db = sns.lmplot(data = supervised4,x=supervised4.columns[3],y=supervised4.columns[4],hue=supervised4.columns[5],fit_reg = False,legend=True,legend_out=True)
```
